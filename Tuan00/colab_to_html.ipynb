# Convert a Jupyter Notebook to HTML using Colab

Steps:
1. Locate your notebook as a `.ipynb`. Colab saved it in your Google Drive.
2. Run the code block below on Colab.
3. Upload your notebook and the converted document will be automatically downloaded. 

```python
try:
    from google.colab import files
    
    f = files.upload()

    # Convert ipynb to html using nbconvert
    import subprocess
    file0 = list(f.keys())[0]
    _ = subprocess.run(["pip", "install", "nbconvert"])
    _ = subprocess.run(["jupyter", "nbconvert", file0, "--to", "html"])

    # download the html
    files.download(file0[:-5]+"html")
except ImportError:
    print("This conversion only works on Google Colab.")

```
